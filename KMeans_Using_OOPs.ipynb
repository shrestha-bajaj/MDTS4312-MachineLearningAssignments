## Import Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Initialize Inputs

```python
X = np.array([
    [1.0,1.0],
    [1.5,2.0],
    [3.0,4.0],
    [5.0,7.0],
    [3.5,5.0],
    [4.5,5.0],
    [3.5,4.5]
])
```

## K-Means

```python
class KMeans:
    def __init__(self, X, K, verbose = True):
        self.X = X
        self.K = K
        self.verbose = verbose

    def select_initial_centroids(self):
        self.centroids = np.random.default_rng().choice(self.X, self.K, replace=False, axis=0)
        if self.verbose:
            print("Starting centroids:", self.centroids)
        return self.centroids

    def closest_centroid(self, point, centroids):
        distances = np.linalg.norm(centroids - point, axis=1)
        return np.argmin(distances)

    def assign_clusters(self):
        idx = []
        for point in self.X:
            cluster = self.closest_centroid(point, self.centroids)
            idx.append(cluster)
        self.idx = np.array(idx)
        return self.idx

    def update_centroids(self):
        new_centroids = []
        for k in range(self.K):
            points = self.X[self.idx == k]
            centroid = np.mean(points, axis=0)
            new_centroids.append(centroid)
        self.new_centroids = np.array(new_centroids)
        return self.new_centroids

    def calculate_cost(self):
        m = len(self.X)
        total = 0
        for i in range(m):
            centroid = self.new_centroids[self.idx[i]]
            total += np.sum((self.X[i] - centroid) ** 2)
        return total / m

    def run_kmeans(self, max_iters=100, plot = True):
        self.select_initial_centroids()

        for i in range(max_iters):
            self.assign_clusters()
            new_centroids = self.update_centroids()
            if np.allclose(new_centroids, self.centroids):
                break
            self.centroids = new_centroids

        if self.verbose:
            print("Resulting centroids:", self.centroids)
            print("Cluster assignments:", self.idx)

        if plot:
            plt.scatter(self.X[:, 0], self.X[:, 1], c=self.idx, s=60, cmap='Set2')
            plt.scatter(self.centroids[:, 0], self.centroids[:, 1], marker='X', color='black', s=200)
            plt.xlabel("X")
            plt.ylabel("Y")
            plt.show()
```

```python
km = KMeans(X, K = 3)
```

```python
km.run_kmeans()
```

## K-Means++

```python
class KMeansPlusPlus(KMeans):
    def select_initial_centroids(self):
        first_idx = np.random.default_rng().choice(len(self.X))
        if self.verbose:
            print("First selected centroid:", self.X[first_idx])
        centroids = [self.X[first_idx]]

        for _ in range(1, self.K):
            centroids_array = np.array(centroids)
            sq_distances = np.array([np.min(np.linalg.norm(centroids_array - point, axis=1))**2 for point in self.X])
            probs = sq_distances / sq_distances.sum()
            next_idx = np.random.default_rng().choice(len(self.X), p=probs)
            centroids.append(self.X[next_idx])

        self.centroids = np.array(centroids)
        return self.centroids
```

```python
KM = KMeansPlusPlus(X, K = 2)
```

```python
KM.run_kmeans()
```

## Elbow Method

```python
def elbow_method(X, k_range=range(2, 6), max_iters=200, method = "Kmeans"):
    costs = []
    for K in k_range:
        if method == "kmeans" or method == "Kmeans" or method == "KMeans" or method == "KMEANS":
            km = KMeans(X, K, verbose = False)
        else:
            km = KMeansPlusPlus(X, K, verbose = False)
        km.run_kmeans(max_iters=max_iters, plot = False)
        cost = km.calculate_cost()
        costs.append(cost)

    plt.plot(list(k_range), costs, marker='s')
    plt.xlabel("K")
    plt.ylabel("Cost")
    plt.show()

    return costs
```

```python
elbow_method(X, method = "kmeans")
```

```python
elbow_method(X, method = "kmeans++")
```
